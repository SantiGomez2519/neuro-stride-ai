Validate that the shared `gait_processing` package loads correctly and that
the processing configuration can be read from `config/processing.yaml`.

| | |

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print("PROJECT_ROOT:", PROJECT_ROOT)
```

```python
from gait_processing import ProcessingConfig

CONFIG_PATH = PROJECT_ROOT / "config" / "processing.yaml"
cfg = ProcessingConfig.from_yaml(CONFIG_PATH)

cfg
```

```python
print("Kinematic Hz:", cfg.kinematic_fs_hz)
print("Analog Hz:", cfg.analog_fs_hz)
print("Marker cutoff:", cfg.marker_cutoff_hz)
print("Normalized points:", cfg.normalized_points)
```

```python
from gait_processing import load_trial

DATA_ROOT = PROJECT_ROOT / "1-experimentation" / "data" / "raw"

trial = load_trial(
    root=DATA_ROOT,
    subject_id="Sub01",
    trial_id="trial0001",
)

trial
```

```python
print(trial.subject_id)
print(trial.trial_id)
print(trial.kinematic_fs_hz)
print(trial.analog_fs_hz)
print(trial.duration_s)
print(trial.mass_kg)
print(trial.height_m)
print(trial.cohort)
print(trial.condition)
```

```python
from gait_processing import check_trial_quality

quality = check_trial_quality(trial)

print("Kinematic integrity:", quality.kinematic_integrity)
print("Analog integrity:", quality.analog_integrity)
```

## Summary

All shared modules load correctly. Configuration is valid. One trial can be loaded and basic quality checks pass.
